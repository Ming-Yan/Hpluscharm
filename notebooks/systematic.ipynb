```python
import matplotlib.pyplot as plt
import uproot
from scipy import stats
import matplotlib.backends.backend_pdf
for y in ["2016_postVFP"]:#["2016_preVFP","2016_postVFP","2017","2018"]:
    for r in ['SR2_LM_1D']:#vars.keys():
        f=uproot.open(f"../card_maker/shape/{y}_{r}_{vars[r]}_newCR.root")
        syst=[i.replace("Up","").replace(";1","") for i in f.keys() if 'ttbar' in i and "Up" in i]
        
        
        
        samples=['hc','ttbar','vv','higgs','st','zjets']
        for s in samples:
            pdf = matplotlib.backends.backend_pdf.PdfPages(f"{y}_{r}_syst_{s}_sep.pdf")
            for sys in f.keys():
                if "Down" in sys: continue
                sys = sys.replace(";1","").replace("Up","").replace("Down","")
                center=f['hc'].axes[0].centers()
                if 'ttbar_weight' in sys and s !='ttbar':continue
                if not sys.startswith(s):continue
                if s==sys:continue
                if "Down" in sys:continue
                fig,ax=plt.subplots()
                m, b, *_ = stats.linregress(center, (f[f'{sys.replace("Up","")}Up'].values()-f[s].values())/f[s].values())
                md, bd, *_d = stats.linregress(center, (f[f'{sys.replace("Down","")}Down'].values()-f[s].values())/f[s].values())
                ax.stairs((f[f'{sys.replace("Up","")}Up'].values()-f[s].values())/f[s].values(),edges=f[s].axes[0].edges(),label='Up')
                ax.stairs((f[f'{sys.replace("Down","")}Down'].values()-f[s].values())/f[s].values(),edges=f[s].axes[0].edges(),label='Down')
                # ax.fill_between(center,
                #y1=,
                # print(m)
                #y2=(f[f'{sys.replace("Up","")}Down;1'].values()-f[s].values())/f[s].values(),alpha=0.2,step='pre',label=s)
                ax.axline(xy1=(0, b), slope=m, label=f'Up:$y = {m:.3f}x+{b:3f}$',color='tab:blue',ls=":")
                ax.axline(xy1=(0, bd), slope=md, label=f'Down:$y = {m:.3f}x+{bd:3f}$',color='tab:orange',ls=":")
                ax.axhline(0.,ls=":",color='gray')
                ax.legend()
                ax.set_title(sys.replace("Up",""))
                # ax.set_ylim(-0.2,0.2)
                ax.set_xlabel("BDT")
                ax.set_ylabel("Relative Uncertainty")
                pdf.savefig(fig)
            pdf.close()
        # rax.fill_between((h.axes[0].edges[1:]+h.axes[0].edges[:-1])/2.,y1=np.sqrt(h.variances())/h.values()+1.,y2=1.-np.sqrt(h.variances())/h.values(),hatch='//',alpha=0.2,facecolor="none",step='pre')
```
